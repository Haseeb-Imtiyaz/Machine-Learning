```python
cat_cols=[]
for i in train.columns:
    if train[i].dtype=='O':
        cat_cols.append(i)
```

```python
for i in cat_cols:
    train[i].fillna(train[i].mode()[0],inplace=True)
```

```python
train.describe()
```

```python
train['Age']=np.where(train['Age']==0,28.827930,train['Age'])
```

```python
train['Age']=train['Age'].fillna(28.827930)
train['RoomService']=train['RoomService'].fillna(0.000000)
train['FoodCourt']=train['FoodCourt'].fillna(0.000000)
train['ShoppingMall']=train['ShoppingMall'].fillna(0.000000)
train['Spa']=train['Spa'].fillna(0.000000)
train['VRDeck']=train['VRDeck'].fillna(0.000000)
```

```python
print(train.isnull().sum()/len(train)*100)
```

```python
def plots(col):
    sns.boxplot(train[col])
    plt.title(col)
    plt.show()
for i in list(train.select_dtypes(exclude=['object'])):
    plots(i)
```

```python
cols=['RoomService','FoodCourt','ShoppingMall','Spa','VRDeck']
for i in cols:
    Q1=train[i].quantile(0.25)
    Q3=train[i].quantile(0.75)
    IQR=Q3-Q1
    upper_limit= Q3 + 1.5 * IQR
    lower_limit= Q1 - 1.5 * IQR
    train[i]=np.where(train[i]>upper_limit,upper_limit,np.where(train[i]<lower_limit,lower_limit,train[i]))
    
def plots(col):
    sns.boxplot(train[col])
    plt.title(col)
    plt.show()
for i in cols:
    plots(i)
```

```python
cols=['HomePlanet','CryoSleep','Destination','VIP','Transported']
for i in cols:
    train[i]=train[i].astype('category')
    train[i]=train[i].cat.codes
train
```

```python
for i in cols:
    print(i)
    print()
    print(train[i].value_counts())
```

```python
X=train.drop('Transported',axis=1)
Y=train['Transported']
```

```python
X.shape,Y.shape
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test= train_test_split(X,Y,test_size=0.2,random_state=1,stratify=Y)
print(X_train.shape,X_test.shape,y_train.shape,y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(X_train,y_train)
```

```python
y_pred_test=lr.predict(X_test)
y_pred_train=lr.predict(X_train)
```

```python
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
print('classification_report of Train: ', classification_report(y_train,y_pred_train),sep='\n\n')
print('classification_report of Test: ', classification_report(y_test,y_pred_test),sep='\n\n')
print('confusion_matrix of Train: ', confusion_matrix(y_train,y_pred_train),sep='\n\n')
print('confusion_matrix of Test: ', confusion_matrix(y_test,y_pred_test),sep='\n\n')
print('accuracy_score of Train: ', accuracy_score(y_train,y_pred_train),sep='\n\n')
print('accuracy_score of Test: ', accuracy_score(y_test,y_pred_test),sep='\n\n')
```
