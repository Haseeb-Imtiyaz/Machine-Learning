##### Many columns have irrelevant data like [Glucose, BloodPressure, BloodPressure  .....  ] have min value = 0 which not possilbe thus updating the dataset
Check outlier by variation in mean, min, max, median etc

    if the column have no outlier change to its mean 
    if it has outlier chnage the data to median
    

```python
df['Glucose']=np.where(df['Glucose']==0, 120.894531, df['Glucose'])
df['BloodPressure']=np.where(df['BloodPressure']==0, 69.105469, df['BloodPressure'])
df['SkinThickness']=np.where(df['SkinThickness']==0, 20.536458, df['SkinThickness'])
df['Insulin']=np.where(df['Insulin']==0, 30.500000, df['Insulin'])   # it has on outlier thus mediann is taken
df['BMI']=np.where(df['BMI']==0, 31.992578, df['BMI'])
```
