```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
```

```python
df=pd.read_csv('German Credit Dataset.csv')
df
```

```python
df.info()
```

```python
for i in df.columns:
    if df[i].dtype=='object':
        df[i]=df[i].astype('category')
        df[i]=df[i].cat.codes
```

```python
df.info()
```

```python
df
```

```python
df.describe()
```

```python
X=df.drop('default',axis=1)
Y=df['default']
```

```python
X
```

## Decison Tree with criterion='gini'

```python
from sklearn.tree import DecisionTreeClassifier
dtree=DecisionTreeClassifier(criterion='gini',min_samples_split=30,min_samples_leaf=10)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test=train_test_split(X,Y,test_size=0.2,stratify=Y)
```

```python
dtree.fit(X_train,y_train)
```

```python
y_pred_train=dtree.predict(X_train)
y_pred_test=dtree.predict(X_test)
```

```python
from sklearn.metrics import classification_report, accuracy_score
print('classification_report test',classification_report(y_test,y_pred_test))
print('classification_report train',classification_report(y_train,y_pred_train))
```

## Descion Tree with criterion='entropy'

```python
dtree=DecisionTreeClassifier(criterion='entropy',min_samples_split=30,min_samples_leaf=10,max_depth=7)
dtree.fit(X_train,y_train)
```

```python
y_pred_train=dtree.predict(X_train)
y_pred_test=dtree.predict(X_test)
from sklearn.metrics import classification_report, accuracy_score
print('classification_report test',classification_report(y_test,y_pred_test))
print('classification_report train',classification_report(y_train,y_pred_train))
```

### Random Forest is used instead of DT

```python
from sklearn.ensemble import RandomForestClassifier
rft=RandomForestClassifier(criterion='entropy',max_depth=6,min_samples_leaf=30)
rft.fit(X_train,y_train)
```

```python
y_pred_rft_train=rft.predict(X_train)
y_pred_rft_test=rft.predict(X_test)
from sklearn.metrics import classification_report, accuracy_score
print('classification_report test',accuracy_score(y_test,y_pred_rft_test))
print('classification_report train',accuracy_score(y_train,y_pred_rft_train))
```
